<h1>Logistic Regression</h1>

```python
import numpy as np
import pandas as pd 
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('daibetes.xlsx')
df.head()
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
df.describe()
```

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import confusion_matrix, accuracy_score, classification_report
```

```python
X = df.drop('Outcome', axis = 1)
y= df['Outcome'].values
print(df.shape, X.shape, y.shape)
```

```python
X_train, X_test, y_train, y_test = train_test_split(X,y, test_size =0.2 , random_state = 32)
```

```python
model = LogisticRegression(max_iter=1000)
```

```python
model.fit(X_train ,y_train)
```

```python
y_pred = model.predict(X_test)
y_pred
```

```python
matrix = confusion_matrix(y_test, y_pred)
acc = accuracy_score(y_test, y_pred)
report = classification_report(y_test, y_pred)
print(matrix)
print(acc)
print(report)
```

## Building linear regression on diabetes data

```python
from sklearn.linear_model import LinearRegression
linreg = LinearRegression()
linreg.fit(X_train, y_train)
```

```python
y_pred_linreg = linreg.predict(X_test)
y_pred_linreg
```

```python
print(y_pred_linreg.max())
print(y_pred_linreg.min())
```

```python
from sklearn.metrics import mean_squared_error, mean_absolute_error, root_mean_squared_error,r2_score
print( mean_squared_error(y_test, y_pred_linreg))
print(mean_absolute_error(y_test, y_pred_linreg))
print( root_mean_squared_error(y_test, y_pred_linreg))
print( r2_score(y_test, y_pred_linreg))
```

```python
pred_list = y_pred_linreg.tolist()
print(pred_list)
```

```python
def sig(x):
    result= 1/(1 + np.exp(-x))
    return result
```

```python
data_for_threshold = []
for i in pred_list:
    data_for_threshold.append(sig(i))
```

```python
print(data_for_threshold)
```

```python
threshold = np.array(data_for_threshold).mean()
print(threshold)
```

```python
output = [1 if x > threshold else 0 for x in data_for_threshold]
print(output)
```

```python
report_manual = classification_report(y_test, output)
print(report_manual)
```
